### Esercizio 3

```python
# Indica il punto di partenza della tabella. La conversione inizierà esattamente da questo numero di chilometri
valore_iniziale = int(input("Inserire il valore iniziale: "))

# Indica il limite massimo da raggiungere. Il programma non deve superare questo valore
valore_finale = int(input("Inserire il valore finale: "))

# Indica l'incremento, cioè di quanto aumenta il valore a ogni riga della tabella. In questo caso, significa che si procede di 5 in 5
valore_passo = int(input("Inserire il passo: "))

if valore_passo <=0:
    print("Il passo deve essere di un valore positivo")

else:
    print(f"{'km':>10} {'miglia':>10} {'metri':>10}")
    print("-" * 33)

km = valore_iniziale
    
while km <= valore_finale:
        miglia = km * 0.621371
        metri = km * 1000
        
        print(f"{km:>10} {miglia:>10.3f} {metri:>10}")
        
        km += valore_passo
```

#### Esercizio 8 con spiegazioni

```python
numero = int(input("Inserire il valore iniziale: "))
if numero < 0:
    print("Il valore deve essere positivo.")
else:
    # Creiamo una stringa vuota per accumulare le cifre binarie. Salviamo il valore in una 
    # variabile temporanea temp
    binario = ""
    temp = numero
    if temp == 0:
        binario = "0"
    while temp > 0:
        # Calcolo e ordine dei resti (binario = str(temp % 2) + binario): L'operatore % 2 
        # calcola il resto della divisione (che può essere solo 0 o 1). Lo convertiamo in 
        # stringa e lo mettiamo prima della stringa esistente (+ binario). Questo passaggio 
        # è cruciale perché il metodo dei resti genera le cifre partendo da quella meno 
        # significativa (a destra) verso quella più significativa (a sinistra).
        binario = str(temp % 2) + binario
        # Aggiornamento del numero (temp = temp // 2): La divisione intera per 2 riduce 
        # il numero a ogni iterazione, avvicinandolo alla fine del ciclo.
        temp = temp // 2
print(f"{numero} in binario: {binario}")
# bin(), calcola automaticamente la rappresentazione binaria di un numero.
# Lo slicing ([2:]) serve ad eliminare "0b" (che da automaticamente la funzione bin())
# per poter leggere solo la rappresentazione binaria, riuscendola a confrontarla
# con la variabile "binario"
if binario == bin(numero)[2:]: 
    print("verifica superata")
else:
    print("verifica fallita")
str_binaria = str(input("inserire una stringa binaria: "))
decimale = 0
# Utilizziamo un ciclo for per scorrere la stringa binaria carattere per carattere,
# procedendo da sinistra verso destra.
for carattere in str_binaria:
    # Verifichiamo che il carattere corrente sia effettivamente un valore binario valido ('0' o '1').
    if carattere != "0" and carattere != "1":
        print("La stringa binaria deve essere composta solo da 0 e 1.")
    else:
        # Moltiplichiamo per 2 il valore accumulato finora (spostando le cifre a sinistra)
        # e sommiamo il valore numerico del carattere corrente (convertito in intero con int()).
        # Questo metodo ricostruisce il valore decimale scorrendo la stringa dall'inizio alla fine.
        decimale = (decimale * 2) + int(carattere)
print(f"{str_binaria} in decimale: {decimale}")
```

## Esercizio 9

```python
aule = "A101;B204;C307;D410;E512"
# .strip() elimina eventuali spazi prima e dopo il testo.
# .upper() trasforma tutte le lettere in maiuscolo.
dato = input("Inserire l'aula che si sta cercando: ").strip().upper()

inizio = 0 # Indica da quale carattere iniziare a leggere la stringa.
contatore = 1 # Conta la posizione dell'aula nell'elenco.
trovato = False # Indica se abbiamo trovato l'aula.

# Il ciclo continua finché non abbiamo raggiunto la fine della stringa.
while inizio < len(aule):
    # In questo caso cerca il punto e virgola ;, partendo dalla posizione
    # indicata da inizio.
    posizione = aule.find(";", inizio) 

    # Quando .find() non trova quello che cerca, restituisce -1
    # Se non trova il separatore, considera come posizione finale la fine della stringa.
    if posizione == -1:
        posizione = len(aule)

    # voce = aule[inizio:posizione] serve a estrarre dalla stringa aule 
    # una parte di testo, partendo dalla posizione inizio fino alla 
    # posizione posizione, esclusa, ottenendo così la singola aula 
    # senza il punto e virgola.
    voce = aule[inizio:posizione]

    # Se sono uguali:
    # Stampiamo il messaggio con la posizione.
    # Mettiamo trovato = True, perché abbiamo trovato l'aula.
    # Usiamo break per interrompere il ciclo, perché non serve continuare a cercare.
    if voce == dato:
        print(f"{dato} è la voce {contatore} dell'elenco.")
        trovato = True
        break

    # Sposta il punto di partenza subito dopo il ;, così il ciclo 
    # può leggere l'aula successiva.
    inizio = posizione + 1
    # Aumenta il contatore di 1.
    # Quindi passiamo dalla prima aula alla seconda, dalla seconda 
    # alla terza e così via.
    contatore += 1

if trovato == False:
    print(f"{dato} non è presente nell'elenco.")
```

## Esercizio 11

```python
frase = input("Inserire la frase: ").lower()  # Legge la frase e trasforma tutte le lettere in minuscolo
palindroma = False  # Inizializza il risultato come falso
ripulita = ""  # Crea una stringa vuota in cui inserire le lettere

frase = frase.replace(" ", "")  # Elimina gli spazi dalla frase

for carattere in frase:  # Esamina un carattere alla volta
    if carattere.isalpha():  # Verifica se il carattere è una lettera
        ripulita += carattere  # Aggiunge la lettera alla stringa normalizzata

print(f"testo normalizzato {ripulita}")  # Stampa la frase senza spazi e segni di punteggiatura

inizio = 0  # Parte dal primo carattere
fine = len(ripulita) - 1  # Parte dall'ultimo carattere; si sottrae 1 perché gli indici partono da 0
confronto = 0  # Inizializza il contatore dei confronti

while inizio < fine:  # Continua finché i due indici non si incontrano
    confronto += 1  # Aumenta il numero dei confronti effettuati

    if ripulita[inizio] == ripulita[fine]:  # Confronta il carattere iniziale con quello finale
        inizio += 1  # Sposta l'indice iniziale verso destra
        fine -= 1  # Sposta l'indice finale verso sinistra
    else:
        break  # Interrompe il ciclo se trova due caratteri diversi

else:
    palindroma = True  # Se il ciclo termina senza interruzioni, la frase è palindroma

print(f"palindroma: {palindroma}")  # Stampa se la frase è palindroma oppure no
print(f"confronti eseguiti: {confronto}")  # Stampa il numero di confronti effettuati

controllo = ripulita == ripulita[::-1]  # Confronta la frase con la sua versione al contrario
print(f"controllo con lo slicing: {controllo}")  # Stampa il risultato della verifica con lo slicing
```

## Esercizio 12

```python
frase = input("Inserire la frase: ")
sottostringa = input("Inserire la sottostringa: ")
if sottostringa == "": # Controlla se la sottostringa è vuota
    print("Sottostringa vuota")
else:
    posizione = 0 # Memorizza la posizione in cui viene trovata la sottostringa
    contatore = 0 # Conta quante volte compare la sottostringa
    inizio = 0 # Indica da quale carattere iniziare la ricerca
    posizioni = []  # Memorizza tutte le posizioni trovate
    while posizione != -1: # Continua finché la sottostringa viene trovata
        posizione = frase.find(sottostringa, inizio)  # Cerca la sottostringa a partire dall'indice indicato
        if posizione != -1: # Controlla se la ricerca ha trovato una corrispondenza
            contatore += 1 # Aumenta il contatore di uno
            posizioni.append(posizione)  # Aggiunge la posizione trovata alla lista
            inizio = posizione + len(sottostringa)
    print("occorrenze:", contatore)
    print("posizioni:", *posizioni)
    print("verifica con count:", frase.count(sottostringa)) # Verifica il numero di occorrenze usando count()
    
    asterischi = "*" * len(sottostringa)  # Crea tanti asterischi quanti sono i caratteri della sottostringa
    mascherata = ""  # Inizializza la nuova frase vuota
    inizio = 0  # Indica da quale posizione riprendere il testo originale

    for posizione in posizioni:  # Scorre tutte le posizioni delle occorrenze trovate
        mascherata += frase[inizio:posizione]  # Aggiunge il testo prima della sottostringa
        mascherata += asterischi  # Sostituisce la sottostringa con gli asterischi
        inizio = posizione + len(sottostringa)  # Sposta l'indice dopo la sottostringa sostituita

    mascherata += frase[inizio:]  # Aggiunge l'eventuale parte finale della frase

    print("testo mascherato:", mascherata)  # Stampa la frase con le occorrenze mascherate
```

### Esercizio 3

```python
# Indica il punto di partenza della tabella. La conversione inizierà esattamente da questo numero di chilometri
valore_iniziale = int(input("Inserire il valore iniziale: "))

# Indica il limite massimo da raggiungere. Il programma non deve superare questo valore
valore_finale = int(input("Inserire il valore finale: "))

# Indica l'incremento, cioè di quanto aumenta il valore a ogni riga della tabella. In questo caso, significa che si procede di 5 in 5
valore_passo = int(input("Inserire il passo: "))

if valore_passo <=0:
    print("Il passo deve essere di un valore positivo")

else:
    print(f"{'km':>10} {'miglia':>10} {'metri':>10}")
    print("-" * 33)

km = valore_iniziale
    
while km <= valore_finale:
        miglia = km * 0.621371
        metri = km * 1000
        
        print(f"{km:>10} {miglia:>10.3f} {metri:>10}")
        
        km += valore_passo
```

### Esercizio 5

```python
frase = "Il Laboratorio 4BI dura 22 ore"

vocali_totali = 0
consonanti_totali = 0
cifre_totali = 0
altri_totali = 0


vocali = "aeiou"
consonanti = "bcdfghljhkmnpqrstvwxyz"
for carattere in frase.lower():
    if carattere.isdigit():
        cifre_totali += 1
    elif carattere in vocali:
        vocali_totali += 1
    elif carattere in consonanti:
        consonanti_totali += 1
    else:
        altri_totali += 1

print(f"vocali: {vocali_totali}, consonanti: {consonanti_totali}, cifre: {cifre_totali}, altri: {altri_totali}")

totale_caratteri = len(frase)
percentuale_vocali = 0.0

if totale_caratteri > 0:
 percentuale_vocali = (vocali_totali / totale_caratteri) * 100
print(f"percentuale di vocali: {percentuale_vocali:.1f}%")
```

### Esercizio 7

```python
numero = 48723

conteggio_cifre = 0
somma_cifre = 0
cifra_maggiore = 0

if numero == 0:
    conteggio_cifre = 1
    somma_cifre = 0
    cifra_maggiore = 0
else:
    temp = numero
    while temp > 0:
        cifra = temp % 10          
        somma_cifre += cifra       
        conteggio_cifre += 1       

        if cifra > cifra_maggiore:
            cifra_maggiore = cifra
            
        temp = temp // 10         

print(f"cifre: {conteggio_cifre}")
print(f"somma delle cifre: {somma_cifre}")
print(f"cifra maggiore: {cifra_maggiore}")
```
